#### Vraag 3b: De oppervlakte lading in het midden van een geleidende schijf op constant potentiaal

1. Met welke formule kan je de oppervlaktelading benaderen in het midden van de schijf($r=0$, $z=0$)?

##### Antwoord 3b

*Geef in deze markdown cel je antwoord op 3b.1.*

````python
import numpy as np

eps0 = 8.8541878128e-12
i_r0 = 0

j_surf = int(np.argmin(np.abs(z - z_pos_disk1)))

Nz = V.shape[1]
if j_surf + 1 < Nz:
    j_out = j_surf + 1
    direction = "net boven"
    Ez_out = -(V[i_r0, j_out] - V1) / grid_step
elif j_surf - 1 >= 0:
    j_out = j_surf - 1
    direction = "net onder"

    Ez_out = +(V[i_r0, j_out] - V1) / grid_step
else:
    raise ValueError("Schijf ligt zo dat er geen punt 'net buiten' bestaat. Vergroot domein of verplaats schijf 1 gridstap naar binnen.")

sigma_mid = eps0 * Ez_out

print(" Benadering oppervlakteladingsdichtheid in het midden schijf")
print(f" Schijfpositie (doel) : z = {z_pos_disk1:.6e} m")
print(f" Gridpunt schijf      : j_surf={j_surf}, z ≈ {z[j_surf]:.6e} m")
print(f" Evaluatiepunt        : {direction} → j_out={j_out}, z ≈ {z[j_out]:.6e} m")
print(f" Ez (net buiten)      : {Ez_out:.6e} V/m")
print(f" σ (met teken)        : {sigma_mid:.6e} C/m²")
print(f" |σ|                  : {abs(sigma_mid):.6e} C/m²")
````

##### Antwoorden 4b
# 1.
We beschouwen twee oneindig grote, parallelle geleidende platen op  
$z = +\frac{d}{2}$ met potentiaal $V_1$  
en  
$z = -\frac{d}{2}$ met potentiaal $V_2$.  
Dus: $\nabla^2 V = 0$.

De Laplace-vergelijking wordt dan:
$\frac{d^2 V}{dz^2} = 0$.
geeft:
$V(z) = Az + B$.

We gebruiken de randvoorwaarden:
$V\left(+\frac{d}{2}\right) = V_1$  
en  
$V\left(-\frac{d}{2}\right) = V_2$.

Invullen:
$A\frac{d}{2} + B = V_1$  
$-A\frac{d}{2} + B = V_2$.

dus
$A = \frac{V_1 - V_2}{d}$.

Het elektrisch veld is:
$\mathbf{E} = - \nabla V$.

Omdat $V$ enkel van $z$ afhangt:
$\mathbf{E} = -\frac{dV}{dz}\,\hat{\mathbf{z}}$.

Met $V(z) = Az + B$ geldt $\frac{dV}{dz} = A$, dus:
$\mathbf{E} = -A\,\hat{\mathbf{z}}$.

geeft:
$\boxed{\mathbf{E} = -\frac{V_1 - V_2}{d}\,\hat{\mathbf{z}}}$.
